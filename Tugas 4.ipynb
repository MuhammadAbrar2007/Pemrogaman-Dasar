```python
#SOAL 1
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_per_buku = int(input("Masukkan harga per buku: "))
uang_bawah = int(input("Masukkan uang pembayaran: "))
total_belanja = jumlah_buku * harga_per_buku
diskon = int(total_belanja * 0.1)  # Diskon 10%
jumlah_bayar = total_belanja - diskon
kembalian = uang_bawah - jumlah_bayar
per_orang = jumlah_bayar // 4
sisa_patungan = jumlah_bayar % 4
uang_cukup = uang_bawah >= jumlah_bayar
print(f"Total       : {total_belanja}")
print(f"Diskon      : {diskon}")
print(f"Bayar       : {jumlah_bayar}")
print(f"Kembalian   : {kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa_patungan}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
#Soal 2
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan rata-rata penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem: "))
kebutuhan_listrik = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))
energi_harian = (jumlah_panel * daya_panel / 1000) * jam_penyinaran * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik
kekurangan = kebutuhan_listrik - energi_harian
mencukupi = energi_harian >= kebutuhan_listrik
print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
#Soal 3
kapasitas_generator = float(input("Masukkan kapasitas generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_bb_per_kwh = float(input("Masukkan konsumsi bahan bakar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus_kabel = float(input("Masukkan arus pada kabel (A): "))
hambatan_kabel = float(input("Masukkan hambatan kabel (Ohm): "))
energi_harian = kapasitas_generator * jam_operasi
kebutuhan_solar = energi_harian * konsumsi_bb_per_kwh
biaya_solar = kebutuhan_solar * harga_solar
hari_cukup = kapasitas_tangki // kebutuhan_solar
sisa_solar = kapasitas_tangki % kebutuhan_solar
rugi_kabel = (arus_kabel ** 2) * hambatan_kabel
rugi_tinggi = (rugi_kabel > 150) and (arus_kabel >= 20)
print(f"Energi (kWh)     : {int(energi_harian)}")
print(f"Solar (liter)    : {float(kebutuhan_solar)}")
print(f"Biaya (Rp)       : {float(biaya_solar)}")
print(f"Tangki cukup     : {float(hari_cukup)} hari")
print(f"Sisa solar (L)   : {float(sisa_solar)}")
print(f"Rugi kabel (W)   : {float(rugi_kabel)}")
print(f"Rugi tinggi?     : {rugi_tinggi}")
```
